### Read all chain samples

```python
from termcolor import colored

chain_probes = [chains_WL, chains_GCsp, chains_3x2pt, chains_3x2pt_GCsp, chains_3x2pt_test, chains_2x2pt_test]
chain_names = ['WL', 'GCsp', '3x2pt', '3x2pt_GCsp', '3x2pt_test', '2x2pt_test']

for i, probe in enumerate(chain_probes):
    print('\n' + colored(f'{"="*10} {chain_names[i]} {"="*10}', 'cyan', attrs=['bold']))
    for chain in probe:
        print(colored(f'→ Working on: {chain}', 'yellow', attrs=['bold']))
        filename = probe[chain]['path']
        with open(filename, 'r') as file:
            header = file.readline().strip().split('\t')
            header[0] = header[0].lstrip('# ')
            data = np.loadtxt(filename, skiprows=1)
            probe[chain]['sample'] = MCSamples(
                samples=data,
                names=header,
                labels=[labels_latex[key] for key in header if key in labels_latex],
                weights=np.exp(data[:, -2]), loglikes=np.exp(data[:, -1])
            )
            probe[chain]['dataframe'] = pd.DataFrame(data=data, index=None, columns=header)
            p = probe[chain]['sample'].getParams()
            p.h = p.H0 / 100
            probe[chain]['dataframe']['h'] = p.H0 / 100
            probe[chain]['dataframe']['10omegam'] = p.omegam * 10
            probe[chain]['dataframe']['10omegab'] = p.omegab * 10
            probe[chain]['sample'].addDerived(p.h, name='h', label='h')
        print(colored(f'✔ Finished: {chain}', 'green'))
print(colored('\nAll chains loaded and processed! 🚀', 'magenta', attrs=['bold']))
```

```python
# Plotting function for individual plots
def plot2D_by_probe(probs, probes_names, models, param1, param2, legend_location='upper right', alp=None, text=[0.07, 0.07], filling=True, lims=None):
    sns.set_style("white")
    for i, prob in enumerate(probs):
        g = plots.get_single_plotter()
        g.settings.axes_fontsize = 14
        g.settings.axes_labelsize = 20
        props = dict(boxstyle='round', facecolor='antiquewhite', edgecolor='black', alpha=0.5)
        plt.text(text[0], text[1], probes_names[i], horizontalalignment='center',
                 verticalalignment='center', transform=plt.gca().transAxes,
                 fontsize=14, bbox=props)
        # Handle filling argument: if it's a bool, make a list of bools for each model
        if isinstance(filling, bool):
            filling_arg = [filling] * len(models)
        else:
            filling_arg = filling
        # Handle alpha argument: if None, use default; if not a list, make a list
        if alp is None:
            alphas_arg = [1.0] * len(models)
        elif isinstance(alp, (float, int)):
            alphas_arg = [alp] * len(models)
        else:
            alphas_arg = alp
        g.plot_2d([prob[mod]['sample'] for mod in models], param1, param2,
                  filled=filling_arg,
                  colors=[prob[mod]['color'] for mod in models],
                  alphas=alphas_arg,
                  shaded=True,
                  lims=lims
                  )
        g.add_x_marker(fiducial[param1])
        g.add_y_marker(fiducial[param2])
        g.add_legend([prob[mod]['legend'] for mod in models], fontsize=10, legend_loc=legend_location)
        plt.tick_params(axis='both', which='both', direction='in', top=True, right=True, left=True, bottom=True)
        g.fig.tight_layout()
        g.fig.savefig('figs/{0}/{1}_{2}_{3}.pdf'.format(probes_names[i], probes_names[i], param1, param2))


# FoM for w0wa
def contour_FoM_calculator(sample, param1, param2, sigma_level=1, names = ['test_name'], colors= ['red', 'blue', 'green']):
    from shapely.geometry import Polygon
    contour_coords = {}
    plt.figure()
    plt.xlabel(r'$w_0$', fontsize=18)
    plt.ylabel(r'$w_a$', fontsize=18)
    names = names
    for ss, i in enumerate(sample):
        #print(names[i])
        density = i.get2DDensityGridData(j=param1, j2=param2, num_plot_contours=3)
        contour_levels = density.contours
        #print(contour_levels)
        contours = plt.contour(density.x, density.y, density.P, sorted(contour_levels),
                               label=names[ss], color = colors[ss]);
        contours.collections[0].set_label(names[ss])
        for ii, contour in enumerate(contours.collections):
            paths = contour.get_paths()
            for path in paths:
                xy = path.vertices
                x = xy[:,0]
                y = xy[:,1]
                contour_coords[ii] = list(zip(x, y))
        sigma_lvls = {3:0, 2:1, 1:2}
        poly = Polygon(contour_coords[sigma_lvls[sigma_level]])  # 0:3sigma, 1:2sigma, 2:1sigma
        area = poly.area
        FoM_area = (2.3*np.pi)/area
        print('FoM {}: {}'.format(names[ss], FoM_area))
        plt.legend();
    plt.tick_params(axis='both', which='major', labelsize=12)
    plt.tick_params(axis='both', which='minor', labelsize=10)
    return FoM_area, density
```
